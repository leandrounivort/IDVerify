Admite un documento y compara el nombre provisto sin margen de error. El nombre debe coincidir exactamente con el del titular del documento (pero no tiene que coincidir con el nombre completo en el documento, sólo la parte provista)

```python
iface = gr.Interface(
    fn=process_image,
    inputs=[
        gr.Image(type="filepath", label="Sube un documento"),
        gr.Textbox(label="Nombres"),
        gr.Textbox(label="Apellidos"),
    ],
    outputs=[
        gr.Textbox(label="Nombres obtenidos"),
        gr.Textbox(label="Apellidos obtenidos"),
        gr.Checkbox(label="Coincide"),
        gr.Checkbox(label="No coincide") 
    ],
    title="Verificador de Nombre en Documento"
)
iface.launch(share=False)
```

Similar a la versión anterior, pero admite la carga de un batch (lote) de documentos para realizar la comprobación de múltiples documentos a la vez.

```python
iface = gr.Interface(
    fn=process_several_images,
    inputs=[
        gr.File(type="filepath", label="Seleccione uno o más documentos",file_count='multiple'),
        gr.Textbox(label="Nombres"),
        gr.Textbox(label="Apellidos"),
    ],
    outputs=[
        gr.Textbox(label="Nombres obtenidos"),
        gr.Textbox(label="Apellidos obtenidos"),
        gr.Checkbox(label="Coincide"),
        gr.Checkbox(label="No coincide")
    ],
    title="Verificador de Nombre en Documentos"
)
iface.launch(share=False)
```

Permite errores de tipeo al comparar el nombre del titular. Se puede configurar el número de errores permitidos, lo que hace que la comparación sea más flexible

```python
iface = gr.Interface(
    fn=process_image_with_misspell,
    inputs=[
        gr.Image(type="filepath", label="Sube un documento"),
        gr.Textbox(label="Nombres"),
        gr.Textbox(label="Apellidos"),
        gr.Number(label="Números de caracteres de tolerancia por palabra",value=0)
    ],
    outputs=[
        gr.Textbox(label="Nombres obtenidos"),
        gr.Textbox(label="Apellidos obtenidos"),
        gr.Checkbox(label="Coincide"),
        gr.Checkbox(label="No coincide"),
    ],
    title="Verificador de Nombre en Documento"
)
iface.launch(share=False)
```
